## Summary
This dataset contains product prices from Amazon UK, with a focus on price prediction. With a good amount of data on what price points sell the most, you can train machine learning models to predict the optimal price for a product based on its features and product name.

### Column Details
1. uid
- Unique identifier for each row. (type: int)
2. asin
- Product ID from Amazon. (type: str)
3. title
- Title of the product. (type: str)
4. stars
- Product rating. If 0, no ratings were found. (type: float)
4. reviews
- Number of customer reviews. If 0, no reviews were found. (type: int)
5. price
- Price of the product. (type: float, Currency: GBP)
6. isBestSeller
- Product category. (type: str)
7. boughtInLastMonth
- Whether the product had the Amazon BestSeller status or not. (type: bool)
8. category
- Indicates how many products we sold last month. (type: int)

```python
# ============================================================
# Amazon UK Cleaned Dataset — Dashboard
# Usage: notebook mein cells ki tarah copy karo, ya poora run karo
# Requires: df = pd.read_parquet("amz_clean.parquet")
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet("amz_clean.parquet")

# ------------------------------------------------------------
# OPTION 1 — STATIC DASHBOARD (matplotlib grid)
# ------------------------------------------------------------
sns.set_style("whitegrid")
fig = plt.figure(figsize=(16, 12))
fig.suptitle("Amazon UK Products — Cleaned Dataset Dashboard", fontsize=16, y=0.98)

# --- 1. Price distribution (log scale) ---
ax1 = fig.add_subplot(2, 3, 1)
sample = df.sample(100_000, random_state=42)          # plotting ke liye sample kaafi hai
sns.histplot(sample["log_price"], bins=60, ax=ax1, color="#2a78d6")
ax1.set_title("log_price distribution (skew = {:.2f})".format(df["log_price"].skew()))
ax1.set_xlabel("log1p(price)")

# --- 2. Rated vs unrated split ---
ax2 = fig.add_subplot(2, 3, 2)
counts = df["has_rating"].value_counts().sort_index()
ax2.bar(["Unrated", "Rated"], counts.values, color=["#c3c2b7", "#1baf7a"])
for i, v in enumerate(counts.values):
    ax2.text(i, v, f"{v/len(df):.1%}", ha="center", va="bottom")
ax2.set_title("Products with vs without ratings")

# --- 3. Stars distribution (rated products only!) ---
ax3 = fig.add_subplot(2, 3, 3)
rated = df[df["has_rating"] == 1]
sns.histplot(rated["stars"], bins=41, ax=ax3, color="#eda100")
ax3.set_title(f"Star ratings (rated only, median = {rated['stars'].median():.1f})")

# --- 4. Top 10 categories by product count ---
ax4 = fig.add_subplot(2, 3, 4)
top_cats = df["category"].value_counts().head(10)
ax4.barh(top_cats.index[::-1].astype(str), top_cats.values[::-1], color="#2a78d6")
ax4.set_title("Top 10 categories by listings")
ax4.tick_params(labelsize=8)

# --- 5. Median price by top categories ---
ax5 = fig.add_subplot(2, 3, 5)
med_price = (df[df["category"].isin(top_cats.index)]
             .groupby("category", observed=True)["price"].median()
             .sort_values())
ax5.barh(med_price.index.astype(str), med_price.values, color="#1baf7a")
ax5.set_title("Median price (top categories)")
ax5.set_xlabel("GBP")
ax5.tick_params(labelsize=8)

# --- 6. Price vs reviews (hexbin, log-log) ---
ax6 = fig.add_subplot(2, 3, 6)
rr = df[(df["reviews"] > 0)].sample(100_000, random_state=42)
ax6.hexbin(rr["log_reviews"], rr["log_price"], gridsize=40, cmap="Blues", mincnt=1)
ax6.set_title("log_price vs log_reviews (density)")
ax6.set_xlabel("log1p(reviews)")
ax6.set_ylabel("log1p(price)")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig("dashboard.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: dashboard.png")

# ------------------------------------------------------------
# OPTION 2 — INTERACTIVE DASHBOARD (Plotly, single HTML file)
# pip install plotly  (agar nahi hai)
# ------------------------------------------------------------
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

s = df.sample(150_000, random_state=42)   # browser ko 2.4M points mat do

fig2 = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Price distribution (log scale)",
        "Top 10 categories",
        "Stars (rated products only)",
        "Median price by category",
    ),
)

fig2.add_trace(go.Histogram(x=s["log_price"], nbinsx=60,
                            marker_color="#2a78d6", name="log_price"),
               row=1, col=1)

tc = df["category"].value_counts().head(10)
fig2.add_trace(go.Bar(y=tc.index.astype(str)[::-1], x=tc.values[::-1],
                      orientation="h", marker_color="#2a78d6", name="listings"),
               row=1, col=2)

fig2.add_trace(go.Histogram(x=df.loc[df["has_rating"] == 1, "stars"], nbinsx=41,
                            marker_color="#eda100", name="stars"),
               row=2, col=1)

mp = (df[df["category"].isin(tc.index)]
      .groupby("category", observed=True)["price"].median().sort_values())
fig2.add_trace(go.Bar(y=mp.index.astype(str), x=mp.values,
                      orientation="h", marker_color="#1baf7a", name="median £"),
               row=2, col=2)

fig2.update_layout(
    title_text="Amazon UK Products — Interactive Dashboard "
               f"({len(df):,} rows, cleaned)",
    height=800, showlegend=False, template="plotly_white",
)
fig2.write_html("dashboard.html")
fig2.show()
print("Saved: dashboard.html  (browser mein kholo, hover/zoom sab chalega)")

# ------------------------------------------------------------
# BONUS — Headline metrics print (dashboard ke upar dikhane ke liye)
# ------------------------------------------------------------
print(f"""
========== HEADLINE METRICS ==========
Rows:            {len(df):,}
Median price:    £{df['price'].median():.2f}
Rated products:  {df['has_rating'].mean():.1%}
Had sales:       {df['had_sales'].mean():.1%}
Categories:      {df['category'].nunique()}
Bestsellers:     {df['isBestSeller'].mean():.2%}
log_price skew:  {df['log_price'].skew():.2f}
======================================
""")
```
