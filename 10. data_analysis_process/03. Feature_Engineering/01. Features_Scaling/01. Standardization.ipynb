### Standardization

xi=xi-mean/standard deviation

```python
# also called z score Standardization
import numpy as np
age=np.array([27,15,33,63,90,0])
standard_deviation=age.std()
mean_age=age.mean()
new_age=[] # new column must have mean=0 and std=1
for i in age:
    new_age.append((i-mean_age)/standard_deviation)
mean_of_resultant_col=sum(new_age)/len(new_age)
print(new_age)
```
