```python
# 'Education_Level','Experience_Level', 'Gender' -> ordinal encoding 
# "Department", "City" ->  nominal encoding-> one hot encoding
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df[['Education_Level','Experience_Level', 'Gender', "Department", "City"]], df["Salary"], test_size=0.3)
x_train.shape, x_test.shape
```

```python
from sklearn.impute import SimpleImputer

cat_imputer = SimpleImputer(strategy="most_frequent")

cat_cols = [
    "Gender",
    "Department",
    "Education_Level",
    "Experience_Level",
    "City"
]

x_train[cat_cols] = cat_imputer.fit_transform(x_train[cat_cols])
x_test[cat_cols] = cat_imputer.transform(x_test[cat_cols])
```

```python
num_imputer = SimpleImputer(strategy="mean")

y_train = num_imputer.fit_transform(y_train.to_frame())
y_test = num_imputer.transform(y_test.to_frame())
```

```python
from sklearn.compose import ColumnTransformer
Transformer=ColumnTransformer(transformers=[
    # ("name",obj,[column names])
    ("tf1",SimpleImputer(strategy="most_frequent"),["Gender","Department","Education_Level","Experience_Level","City"]),
    ("tf2",OrdinalEncoder(categories=[["High School", "Bachelor", "Master", "PhD"],["Junior", "Mid", "Senior"]]),['Education_Level','Experience_Level']),
    ("tf3",OneHotEncoder(sparse_output=False, drop="first"),[ "Department", "City","Gender"])
], remainder="passthrough")
Transformer.fit_transform(x_train).shape
```
