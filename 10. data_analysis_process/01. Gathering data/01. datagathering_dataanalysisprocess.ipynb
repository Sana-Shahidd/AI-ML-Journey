```python
# openning a csv file from url (server se csv file fetch krne ke lia yh code hamesha use hogs)
import requests
from io import StringIO
url ="https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS x 10.15; rv: 66.0) GeoKo/20100101 Firefox/66.0" }
req = requests.get(url, headers=headers)
data= StringIO(req.text)
pd.read_csv(data)
```

Set parameter

```python
# when you are dealing with tsv files
pd.read_csv("data.tsv", sep="/t")
# column ka name khattam hojata hai aur first row col names bn jate hn us case me you can pass names
pd.read_csv("data.tsv", sep="/t", names=["",""])
```

index_col

```python
# when you have id function at the sam time by default index col so remove one 
match=pd.read_csv("matches.csv", index_col="id")
```

Skip bad lines

```python
# id you are facing parser error 8 cols 9 values this will skip rows like this
match=pd.read_csv("matches.csv", nrows=100, error_bad_lines=True)
```

Handling dates

```python
match= pd.read_csv("matches.csv", parse_dates=["date"])
match.info()
# if you want to merge the date column then you have to 
# [[col1, col2]]
```

Converters

```python
# this is for applying trnasformations
match= pd.read_csv("matches.csv", converters={"col_name": "function that you want to apply"})
```

na_values


```python
# I can pass all those values jin ko i want ko nan value consider kia jai for detecting place holders also
pd.read_csv("matches.csv", na_values=["na", "n/a", "not available"])    
```
