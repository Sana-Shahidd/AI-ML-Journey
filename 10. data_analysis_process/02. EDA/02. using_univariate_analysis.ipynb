```python
# when you are doing uni variate analysis the first question that you have to ask whether you are you data of col is numerical or categorical
import pandas as pd
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
# 7 questions that you must ask from each dataset
# 1. datasize
df.shape
# 2. how data looks like
df.sample(5)
# 3. datatypes
df.info()
# 4. Numerical Analysis
df.describe()
# 5. checking you have null values or not if yes exact no
df.isnull().sum()
# 6. checking the duplicate values
df.duplicated().sum()
# 7. checking the covariance
df.corr(numeric_only=True)
```

```python
print(df["Survived"].unique())
```

```python
# in univariate mostly your focus is on plotting the graphs hidden info by data visualization hi milta hai
import seaborn as sns
df["Survived"].value_counts()
sns.countplot(x="Survived", data=df, palette=["red", "green"])
```

```python
df["Survived"].value_counts().plot(kind="pie", autopct="%1.1f%%", colors=["red", "green"], explode=[0, 0.1], shadow=True, startangle=90)
```

2. Numerical columns
* Histogram (creat bins and treat like count plot yh dkene ke lia ke kitne log kitni trf hn to histogram is good option distribution f data in that numerical column)
* Distplot(imprvement f histogram)
* Boxplot

```python
import matplotlib.pyplot as plt
plt.hist(df["Age"], color="blue", edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Age")
plt.show()
```

```python
sns.distplot(df["Age"], color="blue", hist=True, bins=30, )
# the above line the histogram is kde that is used to check the distribution of the data and also to check the skewness of the data this is telling data distribution and it is probability density function and it is used to check the skewness of the data and also to check the distribution of the data
# y axis is probability pdf tells probablity
```

```python
# BOX PLOT( use when you data has noise or outliers)
# IS KA KAM HOTA HAI 5 NUMBER SUMMARY DENA
sns.boxplot(x="Fare", data=df, color="blue")
```
