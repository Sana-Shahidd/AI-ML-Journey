```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
tips = sns.load_dataset("tips")
flight= sns.load_dataset("flights")
iris= sns.load_dataset("iris")
train= pd.read_csv("train.csv")
```

* Numerical - Numerical(Scatter plot, pairplot, line plt)
* Numerical categoricat(Bar plot, box plot, dist plot)
* categorical - categorical(Heatmap, cluster map)

```python
# I want to see the relationship between the two columns 
plt.scatter(x=tips["total_bill"], y=tips["tip"])
plt.xlabel("total_bill")
plt.ylabel("tip")
plt.show()
# bill brh rha hai aur us ke sath tip bhi brh rha hai
```

```python
sns.scatterplot(x=tips["total_bill"], y=tips["tip"], hue=tips["sex"], style=tips["smoker"], size=tips["size"])
# outlirs are male
```

Barplot

```python
titanic=pd.read_csv("Titanic-Dataset.csv")
# pclass ke lehaz se what is the average age of passanger travelling in them
titanic.groupby("Pclass")["Fare"].mean().plot(kind="bar")
```

```python
sns.barplot(x=titanic["Pclass"], y=titanic["Fare"], hue=titanic["Sex"])
```

```python
sns.boxplot(x=titanic["Pclass"], y=titanic["Age"], hue=titanic["Sex"])
```

Distplot (Numerical - categorical) -> probability density function

```python
sns.distplot(x= titanic[titanic["Survived"]==0]['Age'])
sns.distplot(x= titanic[titanic["Survived"]==1]['Age']) # mrne ka probability
```

Heatmap (Categorical - Categorical)

```python
# har pclass me kitne log mar gye aur kitne log zinda han
pd.crosstab(titanic["Pclass"], titanic["Survived"])
```

```python
sns.heatmap(pd.crosstab(titanic["Pclass"], titanic["Survived"]))
```

```python
(titanic.groupby("Pclass")["Survived"].mean() * 100).plot(kind="bar")
```

```python
sns.clustermap(pd.crosstab(titanic["SibSp"], titanic["Survived"]))
# dendogram tells konsi values aik dusre se closer hn
```

Pair plot (ap ne har numerical column ka har numerical column ke sath relation dekhna hai aur scatter plot bnana hai pair plot soves this problem)

```python
sns.pairplot(iris, hue="species") # 3D data ko 2D me plot krna khud ka khud ke sath to different plot bnaya
```

Line plot (Numerical - numerical)

```python
# line based data
# har month me kitne log travel kr rhy hn
flight.groupby("month")["passengers"].mean().plot(kind="line")
```

```python
sns.heatmap(flight.pivot_table(index="month", columns="year", values="passengers"), annot=True, fmt="d")
```
