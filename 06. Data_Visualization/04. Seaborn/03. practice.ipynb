```python
sns.catplot(kind="swarm", data= df, x="Team", y="Sales", hue="Shift")
plt.show()
```

Why do we need Box & Violin plots?

👉 Sometimes average (bar plot) is NOT enough.

You want to know:

How data is spread
Where most values lie
Are there outliers?

👉 That’s where box & violin come in.

📦 1. Box Plot (kind="box")
🎯 What it shows:

👉 5 important things:

Minimum
Q1 (25%)
Median (middle value)
Q3 (75%)
Maximum
Outliers (points outside range)

When to use box plot?

👉 When you want:

Distribution
Outliers
Comparison between groups

2. Violin Plot (kind="violin")
🎯 What it shows:

👉 Everything box plot shows + distribution shape

👉 It tells:

Where data is dense
Where data is sparse
👉 Box = summary
👉 Violin = summary + shape

```python
import pandas as pd

data = {
    "Team": ["A","A","B","B","A","B","A","B","A","B"],
    "Sales": [200,220,250,270,300,320,350,370,400,420],
    "Profit": [20,25,30,35,40,45,50,55,60,65],
    "Shift": ["Morning","Evening","Morning","Evening","Morning","Evening","Morning","Evening","Morning","Evening"]
}

df = pd.DataFrame(data)
```

```python
sns.catplot(kind="box", data=df, x="Team", y="Sales", hue="Shift")
plt.show()
```

```python
sns.catplot(kind="violin", data=df, x="Team", y="Profit", hue="Shift", split=True)
plt.show()
```

Pairplot and pairgrids

```python
import pandas as pd

data = {
    "Team": ["A","A","B","B","A","B","A","B","A","B"],
    "Sales": [200,220,250,270,300,320,350,370,400,420],
    "Profit": [20,25,30,35,40,45,50,55,60,65],
    "Shift": ["Morning","Evening","Morning","Evening","Morning","Evening","Morning","Evening","Morning","Evening"]
}

df = pd.DataFrame(data)
```

pairplot->readymate
diagnol->histogram
offset->scatter

```python
sns.pairplot(data=df, hue="Shift")
```

```python
g=sns.PairGrid(data=df, hue="Shift")
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_upper()->above diagnol
# map_lower()->below diagnol
```

```python
# What is vars in PairGrid?

# 👉 vars = list of columns you want to include in the grid
```

```python
g = sns.PairGrid(data=df,hue='Shift',vars=['Sales','Team'])
g.map_diag(sns.histplot)
```

### JointGrid vs Jointplot

Shows:

scatter (center)
histograms (top & side)

```python
sns.jointplot(data=df, x="Sales", y="Profit")
plt.show()
```

```python
sns.jointplot(data=df, x="Sales", y="Profit", kind="reg")
```

```python
sns.jointplot(data=df, x="Sales", y="Profit", kind="kde")
```

```python
g = sns.JointGrid(data=df, x="Sales", y="Profit")

g.plot_joint(sns.scatterplot)   # main plot
g.plot_marginals(sns.histplot)  # top & side plots

plt.show()
```
