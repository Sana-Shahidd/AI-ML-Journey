```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    "Student": ["A","B","C","D","E","F","G","H","I","J"],
    "Math": np.random.randint(50, 100, 10),
    "Physics": np.random.randint(50, 100, 10),
    "Chemistry": np.random.randint(50, 100, 10),
    "Hours_Studied": np.random.randint(1, 10, 10),
    "Attendance": np.random.randint(60, 100, 10),
    "Age": np.random.randint(18, 25, 10)
}

df = pd.DataFrame(data)
print(df)
```

# Line Plot

```python
# * Q1. Plot a line chart showing Math, Physics, and Chemistry scores for all students.
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.plot(df["Student"], df["Math"], label="Maths", marker="o")
plt.plot(df["Student"], df["Physics"], label="Physics", marker="o")
plt.plot(df["Student"], df["Chemistry"], label="Chemistry", marker="o")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Performance Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# * Q2. Plot a line graph of Hours_Studied vs Math marks. #better approch is to use scatter plot but as per question we will use line graph
plt.figure(figsize=(10,6))
plt.plot(df["Hours_Studied"], df["Math"], marker="o")
plt.xlabel("Hours_Studied")
plt.ylabel("Maths")
plt.title("Hours studied vs Maths")
plt.grid(True)
plt.show()
```

Bar Plot->for displaying the categorical data and univariate analysis x_axis->categorical data, y_axis->height

```python
# Q3. Create a bar plot showing Math marks of each student.
plt.figure(figsize=(10,6))
plt.bar(df["Student"], df["Math"], color="skyblue", edgecolor="black", linewidth=1.5, alpha=0.7)
plt.xlabel("Students")
plt.ylabel("Math Marks")
plt.title("Math Marks of Each Student")
plt.grid(axis="y")
plt.show()
```

```python
# Q4. Create a grouped bar chart comparing Math, Physics, and Chemistry for each student.
plt.figure(figsize=(10,6))
plt.bar(np.arange(df.shape[0])+0.25, df["Math"], width=0.25)
plt.bar(np.arange(df.shape[0]), df["Physics"], width=0.25)
plt.bar(np.arange(df.shape[0])-0.25, df["Chemistry"], width=0.25)
plt.xticks(np.arange(df.shape[0]), df["Student"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Performance Comparison")
plt.legend(["Math", "Physics", "Chemistry"])
plt.grid(axis='y')
plt.show()
```

Histogram-> univariate analysis, numerical analysis

```python
# Q5. Plot a histogram of Math scores distribution.
plt.figure(figsize=(8,5))
plt.hist(df["Math"], bins=10, color="lightcoral", edgecolor="black", linewidth=0.5)
plt.xlabel("Math Scores")
plt.ylabel("Frequency")
plt.title("Distribution of Math Scores")
plt.grid(axis='y')
plt.show()
```

```python
# Q6. Plot histograms for all three subjects in one figure (use transparency).
plt.figure(figsize=(10,6))
plt.hist(df["Math"], bins=5, color="lightblue", alpha=0.5, label="Math")
plt.hist(df["Physics"], bins=5, color="darkblue", alpha=0.5, label="Physics")
plt.hist(df["Chemistry"], bins=5, color="lightgray", alpha=0.5, label="Chemistry")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Distribution of Subject Scores")
plt.legend()
plt.grid(axis='y')
```

## Scatter plots



```python
# Q7. Create a scatter plot of Hours_Studied vs Math marks.
plt.figure(figsize=(10,6))
plt.scatter(df["Hours_Studied"], df["Math"], marker="o")
plt.xlabel("Hours_Studied")
plt.ylabel("Maths")
plt.title("Hours studied vs Maths")
plt.grid(True)
plt.show()
```

```python
# Q8. Create a scatter plot of Attendance vs Physics marks and add student names using plt.text().
plt.figure(figsize=(10,6))
plt.scatter(df["Attendance"], df["Physics"], marker="o")
plt.xlabel("Attendance")
plt.ylabel("Physics")
plt.title("Attendance vs Physics")
for i in range(len(df)):
    plt.text(df["Attendance"].iloc[i], df["Physics"].iloc[i], df["Student"].iloc[i], fontsize=8)
plt.grid(True)
plt.show()
```

Piechart

```python
# Q9. Show the average marks distribution of 3 subjects using a pie chart.
plt.figure(figsize=(3,3))
plt.pie([df["Math"].mean(), df["Physics"].mean(), df["Chemistry"].mean()], labels=["Math", "Physics", "Chemistry"], autopct="%1.1f%%")
plt.title("Average Marks Distribution")
plt.show()
```

## 3D Surface Plot

```python
# Q15. Create a surface plot using:

X, Y = np.meshgrid(range(10), range(10))
Z = np.sin(X) + np.cos(Y)
fig=plt.figure(figsize=((5,5)))
ax= plt.subplot(projection="3d")
p= ax.plot_surface(X,Y,Z, cmap="viridis")
fig.colorbar(p)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

xx, yy = np.meshgrid(range(10), range(10))
zz = xx**2 + yy**2

fig = plt.figure(figsize=(6,5))
ax = fig.add_subplot(projection='3d')

surface = ax.plot_surface(xx, yy, zz, cmap="plasma")

ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("3D Surface Plot: z = x² + y²")

fig.colorbar(surface)
# how to save this as image
plt.savefig("surface_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```
