```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("sharma-kohli.csv")
df
```

## Line Plot

* bivariate analysis
* continuous data
* usually for time series data

```python
plt.plot(df["index"], df["RG Sharma"], linewidth=2, color="red", marker="o", markersize=10, markerfacecolor="blue")
plt.plot(df["index"], df["V Kohli"], label="year vs player runs", linewidth=2, color="blue", marker="o", markersize=10, markerfacecolor="red")
plt.xlabel("years")
plt.ylabel("runs")
plt.legend()
plt.show()
```

```python
plt.Figure(figsize=(10,5))
plt.subplot(1, 2, 1)
plt.plot(df["index"], df["RG Sharma"], linestyle="--",linewidth=2, color="red", marker="o", markersize=10, markerfacecolor="blue")
plt.title("Rohit Sharma")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(df["index"], df["V Kohli"], linewidth=2, color="blue", marker="o", markersize=10, markerfacecolor="red")
plt.title("Virat Kohli")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.tight_layout()
plt.grid(True)
plt.show()
```

## Bar plot

* for categorical data
* bivariate analysis one is category and other is numerical

```python
df= pd.read_csv("iris.csv")
plt.bar(df["Species"], df["PetalLengthCm"], color="red", width=0.5, edgecolor="blue", label="Petal Length vs Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.title("Iris Petal Length by Species")
plt.xticks(rotation=90)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
fig, ax= plt.subplots(1, 2, figsize=(10, 5))
ax[0].bar(df["Species"], df["PetalLengthCm"], color="red", width=0.5, edgecolor="blue", label="Petal Length vs Species")
ax[0].set_xlabel("Species")
ax[0].set_ylabel("Petal Length (cm)")
ax[0].set_title("Iris Petal Length by Species")

ax[1].bar(df["Species"], df["PetalWidthCm"], color="blue", width=0.5, edgecolor="red", label="Petal Width vs Species")
ax[1].set_xlabel("Species")
ax[1].set_ylabel("Petal Width (cm)")
ax[1].set_title("Iris Petal Width by Species")
plt.show()
```

Histogram

* distribution of numerical data
* how values are spread
* Instead of categories → it groups values into bins (ranges)

```python
data = [10, 20, 20, 30, 30, 30, 40, 50]
plt.hist(data, bins=10,
         color="skyblue",
         edgecolor="black", histtype="bar", rwidth=0.8)

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.scatter(
    x, y,
    color="blue",        # color of points
    s=100,               # size of points
    alpha=0.7,           # transparency
    marker="o",          # shape of marker
    edgecolor="black",   # border color
    linewidth=1,         # border thickness
    label="Data Points"  # legend label
)

plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()
plt.grid(True)

plt.show()
```
