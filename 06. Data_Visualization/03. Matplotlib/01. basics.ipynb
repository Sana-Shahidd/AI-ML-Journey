```python
batsman= pd.read_csv("sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman["index"], batsman["V Kohli"], 
         color= "green",
         linestyle="solid",
         linewidth=3,
         marker="D",
         markersize=4,
         label="kholi",
        )#+,.,>,O
plt.legend(loc="best")
```

```python
#plotting multiple plots
plt.plot(batsman["index"], batsman["V Kohli"])
plt.plot(batsman["index"], batsman["RG Sharma"])
plt.xlabel("year")
plt.ylabel("scores")
plt.title("career comparison")
plt.grid()
plt.show()
```

```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc='upper right')
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,75000)
plt.xlim(2017,2019)
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()
plt.show()
```

# bar chart

```python
# x-axis->categoricat
# y-axis->aggregate operation perform krte ho
# numerical vs categorical
children=[10, 20, 25, 5]
colours=["red","blue","green","yellow"]
plt.bar(colours, children)
```

```python
plt.barh(colours, children) #horizontal bar chart more of preference thing no of categories increases use horizontal
```

```python
#labelling multiple quantities
df=pd.read_csv("batsman_season_record.csv")
plt.bar(np.arange(df.shape[0])-0.3,df["2015"], width=0.3)
plt.bar(np.arange(df.shape[0]),df["2016"], width=0.3)
plt.bar(np.arange(df.shape[0])+0.3,df["2017"], width=0.3)
plt.xticks(np.arange(df.shape[0]), df["batsman"])
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
```

```python
#scattered barchart
plt.bar(df['batsman'], df['2017'], label='2017')
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], label='2016')
plt.bar(df['batsman'], df['2015'], bottom=df['2017'] + df['2016'], label='2015')

plt.legend()
plt.show()
```

```python
import pandas as pd

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Zara"],
    "Math": [78, 85, 90, 88],
    "Physics": [72, 80, 85, 83],
    "CS": [88, 92, 95, 91]
}

df = pd.DataFrame(data)
df
```

```python
np.arange(df.shape[0])
```

```python
#bar chart means categorical data on x-axis 
plt.bar(np.arange(df.shape[0])+0.3, df["Math"] , width=0.3)
plt.bar(np.arange(df.shape[0]), df["Physics"], width=0.3)
plt.bar(np.arange(df.shape[0])-0.3, df["CS"], width=0.3)
plt.xticks(np.arange(df.shape[0]), df["Student"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.legend(["Math", "Physics", "CS"])
plt.show()
```
