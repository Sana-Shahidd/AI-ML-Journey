Data Cleaning

```python
#  Count missing values in the breed column.
Cats["breed"].isna().sum()

# Replace empty strings with NaN.
#No empty string in data
Cats["breed"].replace(" ", np.nan, inplace=True)

# Fill missing values in breed with "Unknown".
#There is no missing value in breed column
#otherwise this would be syntax
Cats["breed"].fillna("Unknown", inplace=True)

# Fill missing weight_kg with the mean weight.
Cats[Cats["weight_kg"] <= 0]   # 9 rows
median_weight=Cats.loc[Cats["weight_kg"] > 0, "weight_kg"].median()
Cats.loc[Cats["weight_kg"]<=0, "weight_kg"] = median_weight

# Remove rows where age_years is negative.
# Cats.dropna(Cats["age_years"]<0) #dropna works for only missing values not conditions
Cats = Cats[Cats["age_years"] >= 0] 

# Remove rows where weight_kg is greater than 20 kg.
# Cats.dropna(Cats["weight_kg"]>20) #same reason as above
Cats = Cats[Cats["weight_kg"] <= 20]

# Standardize gender values (Male, male, M → Male).
Cats["gender"].value_counts()
Cats["gender"] = Cats["gender"].str.lower()
standarized_male={
"male":"Male",
"m":"Male",
"female":"Female",
"f":"Female"
}
Cats["gender"].replace(standarized_male, inplace=True)

# Convert all breed values to lowercase also Remove leading/trailing spaces in breed.
Cats["breed"]=Cats["breed"].str.lower().str.strip()
Cats["breed"].unique()

# Replace "unknown" values with NaN.
Cats["breed"].replace("Unknown", np.nan, inplace=True)

# Drop rows where name is missing.
Cats=Cats[~Cats["name"].isna()]
#other way
Cats.dropna(subset=["name"], inplace=True)

# Fill missing diet_type with mode.
mode_diet = Cats["diet_type"].mode()[0]  # mode() returns a Series, take the first value
Cats["diet_type"].fillna(mode_diet, inplace=True)

# Convert last_vet_visit column to datetime.
Cats["last_vet_visit"] = pd.to_datetime(Cats["last_vet_visit"], errors="coerce")

# Handle invalid dates in last_vet_visit.
Cats["last_vet_visit"].fillna(pd.Timestamp.today(), inplace=True)

# Convert numeric columns stored as strings into numbers.
# List of numeric columns to convert
numeric_cols = ["weight_kg", "age_years"]  # add other numeric columns if needed

# Convert each column to numeric, invalid entries become NaN
for col in numeric_cols:
    Cats[col] = pd.to_numeric(Cats[col], errors="coerce")

# Check if any column contains only missing values.
missing_only_cols = Cats.columns[Cats.isna().all()]

# Drop columns with more than 60% missing values.
numeric_cols = ["weight_kg", "age_years"]  # add other numeric columns as needed
for col in numeric_cols:
    Cats[col] = pd.to_numeric(Cats[col], errors="coerce")  # invalid strings become NaN

# Remove duplicate rows again after cleaning.
Cats.drop_duplicates(inplace=True)

# Verify the dataset shape after cleaning.
Cats.shape
```

Sorting & Ranking

```python
# Sort dataset by age_years.
Cats.sort_values("age_years")

# Sort dataset by weight_kg descending.
Cats.sort_values("weight_kg", inplace=True)

# Sort by breed alphabetically.
Cats.sort_values("breed")

# Find top 10 heaviest cats.
Cats.sort_values("weight_kg", ascending=False).head(10)

# Find 10 youngest cats.
Cats.sort_values("age_years").head(10)

# Rank cats by weight.
Cats["weight_rank"] = Cats["weight_kg"].rank(ascending=False)

# Rank cats by age.
Cats["age_rank"] = Cats["age_years"].rank(ascending=True)

# Find top 5 breeds with most cats.
Cats.groupby("breed").size().sort_values(ascending=False).head(5)

# Find top 5 cities with most cats.
Cats.groupby("owner_city").size().sort_values(ascending=False).head(5)

# Sort dataset by city and age.
Cats.sort_values(["owner_city","age_years"])
```

Feature Engineering

```python
# Create a column age_in_months.
Cats["age_in_months"]=Cats["age_years"].apply(lambda x: x*12)
Cats.shape

# Create a column weight_in_grams.
Cats["weight_in_grams"]=Cats["weight_kg"].apply(lambda x: x*1000)
Cats.shape

# Create a column is_senior_cat (age > 10).
Cats["is_senior_cat"]=Cats["age_years"].apply(lambda x: x>10)

# Create weight_category (Underweight / Normal / Overweight).
mean_weight=Cats["weight_kg"].mean()
Cats["weight_category"]=Cats["weight_kg"].apply(lambda x: "Underweight" if x<mean_weight else("Overweight" if x>mean_weight else "Normal"))

# Create age_group (Kitten / Adult / Senior).
mean_age=Cats["age_years"].mean()
Cats["age_group"]=Cats["age_years"].apply(lambda x: "kitten" if x<mean_age else("senior" if x>mean_age else "adult"))

# Extract year from last_vet_visit.
Cats["last_year"] = Cats["last_vet_visit"].dt.year

# Create a column name_length.
Cats["name_length"]=Cats["name"].str.len()

# Create column is_heavy_cat (weight > 6).
Cats["is_heavy_cat"] = Cats["weight_kg"] > 6

# Create column is_indoor_cat randomly.
Cats["is_indoor_cat"] = np.random.choice([True, False], size=len(Cats))

# Create column cat_score = age * weight.
Cats["cat_score"] = Cats["age_years"] * Cats["weight_kg"]
```
