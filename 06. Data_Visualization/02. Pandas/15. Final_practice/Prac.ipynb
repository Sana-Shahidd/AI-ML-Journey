### function to remove the dirty  or missing values other than nan from whole data frame.

```python
def remove_missing_values(df):
        df=df.replace([r"(?i)^\s*(unknown|missing|null|none|n/a)|\?\s*$"], np.nan, regex=True)
        return df
#regex=True allows Pandas to interpret the pattern as a regular expression, enabling flexible matching instead of exact string replacement.
```

### function for strandardizing the column names

```python
def standardize_col_names(df):
    df.columns=(
            df.columns
            .str.lower()
            .str.strip()
            .str.replace(" ","_"))
    return df
```

```python
item=standardize_col_names(item)
item.columns
```

### if 60 percent of data is missing in column better to drop

```python
def threshhold(df):
    thresh=0.60*len(df)
    columns_to_drop=df.columns[df.isna().sum()>threshhold]
    df = df.drop(columns= columns_to_drop)
    return df
```

### drop the duplicate columns

```python
def drop_duplicate_columns(df):
    df=df.loc[: , ~df.columns.duplicated()]
    return df
```

```python
item=drop_duplicate_columns(item)
item.shape
```

```python
def drop_duplicate_rows(df):
    df=df.drop_duplicates()
    return df
```

```python
item=drop_duplicate_rows(item)
item.shape
```

### if i want to replace the numerical columns missing values then  calculate the mean of of non negative values and fill in the missing columns

```python
# for numerical columns
def replace_missing_with_mean(df):
    for col in df.select_dtypes(include="number").columns:
        mean_val=df.loc[df[col]>0, col].mean()
        df[col] = df[col].fillna(mean_val)
    df[col] = df[col].fillna(mean_val)
```

```python
# for categorical columns use mode
def replace_with_mode(df):
    for col in df.select_dtypes(include="object").columns:
        if not df[col].mode().empty:
            mode_val = df[col].mode()[0]
            df[col] = df[col].fillna(mode_val)
    return df
```

```python
item=replace_with_mode(item)
item["product_name"].isna().sum()
item.info()
```

### strandardize the textual values

```python
def strandardize_text_data(df):
    for col in df.select_dtypes(include="object").columns:
        df[col]=(df[col]
                .str.strip()
                .str.lower()
                )
    return df
```

```python
import pandas as pd
import numpy as np

def clean_price_columns(df):
    df = df.copy()

    cols = ["list_price", "standard_cost"]

    for col in cols:
        if col in df.columns:
            df[col] = (
                df[col]
                .astype(str)
                .str.strip()
                .str.lower()
                # remove currency text
                .str.replace("usd", "", regex=False)
                .str.replace("$", "", regex=False)
                # remove commas (1,200 → 1200)
                .str.replace(",", "", regex=False)
                # replace invalid words with NaN
                .replace(["nan", "none", "null", "n/a", "na", "?", "error", ""], np.nan)
            )

            # convert to numeric
            df[col] = pd.to_numeric(df[col], errors="coerce")

    df.loc[df["list_price"] < 0, "list_price"] = np.nan
    df.loc[df["standard_cost"] < 0, "standard_cost"] = np.nan

    # cost should not exceed price
    df.loc[df["standard_cost"] > df["list_price"], "standard_cost"] = np.nan

    return df
```

```python
item = clean_price_columns(item)
```
