```python
import pandas as pd
import numpy as np
import matplotlib.pyplot
```

```python
dogs=pd.read_csv("dogs_dirty_dataset.csv")
```

```python
dogs["weight_kg"] = pd.to_numeric(dogs["weight_kg"], errors="coerce")
dogs["height_cm"] = pd.to_numeric(dogs["height_cm"], errors="coerce")
dogs["owner_age"] = pd.to_numeric(dogs["owner_age"], errors="coerce")
dogs["daily_walk_minutes"] = pd.to_numeric(dogs["daily_walk_minutes"], errors="coerce")
dogs["barks_per_day"] = pd.to_numeric(dogs["barks_per_day"], errors="coerce")
dogs["vet_visits_per_year"] = pd.to_numeric(dogs["vet_visits_per_year"], errors="coerce")
```

```python
#First i have to check that is there any column that is taking unnecessary space
dogs.info()
#dogs id can be stored in int8 rather than int64 because these are just 4 integers
dogs["dog_id"]=dogs["dog_id"].astype("Int32")
#same for age_years
dogs["age_years"]=dogs["age_years"].astype("Int8")
#weight and integer should be float or integer not object same for owners age
dogs["weight_kg"]=dogs["weight_kg"].astype("float32")
dogs["height_cm"]=dogs["height_cm"].astype("float32")
dogs["owner_age"]=dogs["owner_age"].astype("float16")
dogs["owner_income"]=dogs["owner_income"].astype("float16")
dogs["daily_walk_minutes"]=dogs["daily_walk_minutes"].astype("Int16")
dogs["barks_per_day"]=dogs["barks_per_day"].astype("Int16")
dogs_cols=[
    "breed","color","diet_type","training_level",
"owner_city","vaccinated","neutered"
]
for col in dogs_cols:
    dogs[col] = dogs[col].astype("category")
dogs["last_vet_visit"] = pd.to_datetime(dogs["last_vet_visit"], errors="coerce")
dogs["adoption_date"] = pd.to_datetime(dogs["adoption_date"], errors="coerce")
sensor_cols = [col for col in dogs.columns if "sensor_metric" in col]
for col in sensor_cols:
    dogs[col] = pd.to_numeric(dogs[col], errors="coerce").astype("float32")
```

```python
#strandaize the column names
dogs.columns=(dogs.columns.
              str.strip().
              str.lower().
              str.replace(" ","_"))
```

```python
#remove the duplicate rows
dogs.drop_duplicates(inplace=True)
```

```python
# Handle impossible values
dogs.loc[dogs["weight_kg"] <= 0, "weight_kg"] = np.nan
dogs.loc[dogs["height_cm"] <= 0, "height_cm"] = np.nan
dogs.loc[dogs["age_years"] < 0, "age_years"] = np.nan
```

```python
# Missing value summary
missing = dogs.isna().sum().sort_values(ascending=False)
print(missing)
```

```python
#Drop the columns where too many missing values
threshold=len(dogs) * 0.60
dogs = dogs.dropna(thresh=threshold, axis=1)
```
