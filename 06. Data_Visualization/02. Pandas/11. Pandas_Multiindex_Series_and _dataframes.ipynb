```python
#can we make multi index
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
a = pd.Series([1,2,3,4,5,6,7,8],index=index_val)
# a([])
```

```python
# correct way
#multi indexing series uses multilevel indexing also known as hierarchal indexing
# For creating there are two ways
# 1.pd.MultiIndex.from_tuples()
multiindex=pd.MultiIndex.from_tuples(index_val)
multiindex.levels #decouples
multiindex.levels[0]
multiindex.levels[1]
# 1.pd.MultiIndex.from_product()
pd.MultiIndex.from_product([['cse', 'ece'], [2019, 2020, 2021, 2022]]) #formed by cartesian prod
```

```python
s=pd.Series([1,2,3,4,5,6,7,8], index=multiindex).reset_index()
s
```
