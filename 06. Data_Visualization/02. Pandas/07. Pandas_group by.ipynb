GroupBy
make groups on the basis of some column 
Types of column:
Numerical
categorical
group by applies on actegorical data

```python
import pandas as pd
import numpy as np
```

```python
movies=pd.read_csv('imdb-top-1000.csv')
movies
```

```python
ans=movies.groupby(["Genre","Director"])
ans.sum()
```

```python
genres=movies.groupby('Genre')
```

```python
#apply builtin aggregation function on groupby objects
genres.sum()
# type(genres.sum()) #data frame
```

```python
genres.min()
```

```python
#find the top 3 genres by total earning
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
#find the top 3 genres by total earning
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
# Find the genre with highest imdb rating
movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
#find the director with most popularity on the basis of no of votes
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# highest rated movie on each genre on basis of imdb rating
highest_rating=movies.groupby('Genre')['IMDB_Rating'].max()
movies
```

```python
#find no of movies done by each actor
movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
# movies['Star1'].value_counts().head(1)
```

#Groupby attributes and methods
#find total no of grps->len
#items in each group->size
#first()/last()->nth item
#get_group() all movies of that group
#groups
#describe
#sample
#nunique return no of unique values

```python
len(movies.groupby('Star1'))
movies['Star1'].nunique()
```

```python
movies.groupby('Star1').size() #by default sort on basis of index
movies['Star1'].value_counts() #by default sort on basis of values
```

```python
genres.first() 
genres.last() 
genres.nth(7) #0 based indexing
```

```python
genres.get_group('Horror')
```

```python
genres.groups #dictionary with index position of that genre
```

```python
genres.sample(2,replace=True)
```
