### program-> 
* program is sequence of instructions written in programming language for instance gc(entire application)->exe-> program written and it tell how particular browser should work

### process->  
* A process is simply an instance of a program that is being executed e.g gc double click gc open/ resources are shared within the process so works with os code segment/ data segment(list of global and static variables)/ heap/ stack(access of functions)/ resistor

* every process will have a seperate memory space like for 1-> sep space.
* one process cannot corrupt another space cuz of sep memory space 
* cuz of sep space I/O requirement -> increased execution time to switch between the processes

### Thread ->
* unit of execution with in a process
* types:
1-> single threaded process: own stack and resistor but sharing CS(code segment), DS(data segment), Heap for instance MS paint threads creation for doing multiple tasks like creation of shapes
2-> Multithreaded process: both threads have own stack and registor but sharing CS/DS/heap


### Why and when to use
* I/O bounded tasks: tasks that spend more on timing operations (e.g, file operations)
* Concurrent operations-> When you want to improve the throughput of your application

Example of multithreading

```python
import threading
import time
```

```python
def print_numbers():
    for i in range(5):
        time.sleep(2) # I/O operation happening
        print(f"Numbers:{i}")


def print_letter():
    for letter in "abcde":
        time.sleep(2)  # I/O operation happening
        print(f"Letter: {letter}")

#creating 2 threads
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)
```

```python
t=time.time()
#instead if calling the functions just start the thread
t1.start()
t2.start()
# before returning wait for the threads to join
t1.join()
t2.join()

finished_time=time.time()-t
print(finished_time)
```

```python
t=time.time()
print_numbers()
print_letter()
finished_time=time.time()-t
print(finished_time)
```

### MultiThreading with thread pool executor

```python
from concurrent.futures import ThreadPoolExecutor
import time

def print_numbers(numbers):
    time.sleep(1)
    return f'Numbers: {numbers}'

numbers=[1,2,3,4,2,3,4,5,6,7,8,9,0]

with ThreadPoolExecutor(max_workers=3) as executor:
    results=executor.map(print_numbers, numbers)

for result in results:
    print(result)
```
