```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def div(x, y):
    try:
        result = x / y
        logging.info(f"Division of {x} by {y} is {result}")
        return result

    except ZeroDivisionError as e:
        logging.error("Division failed")


a = div(10, 0)
print(a)
```

```python
import logging

logging.basicConfig(
    filename="myfile.log",
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

def factorial(n):
    logging.info(f"Started factorial of {n}")
    
    result = 1
    for i in range(1, n + 1):
        result *= i
        logging.debug(f"i={i}, result={result}")
    
    logging.info(f"Finished factorial of {n}")
    return result

print(factorial(5))
```
