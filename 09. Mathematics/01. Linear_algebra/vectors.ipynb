## Eucledian's Distance for recomendation system

applying the eucledians distnce to check the similarity this is used in k nearest neighbour. 

```python
import numpy as np
x=np.array([1,2,3,4,5])
y=np.array([4,5,6,7,8])
# now calculate the distance
distance=(x-y)**2
eucledian_distance=0
for i in distance:
    eucledian_distance=i+eucledian_distance
result=(eucledian_distance)**0.5
print(result)
```

```python
import numpy as np

# Ratings of users
user1 = np.array([5, 3, 4])
user2 = np.array([4, 3, 5])
user3 = np.array([1, 5, 2])

# Euclediean Distance
distance12=np.linalg.norm(user1 - user2)
distance13=np.linalg.norm(user1 - user3)

print("Distance(User1, User2):", distance12)
print("Distance(User1, User3):", distance13)
```

Scalar Addition and Subtraction


```python
import numpy as np
import matplotlib.pyplot as plt

# Original data
x = np.array([1, 2, 3, 4, 5])

# Scalar addition
x_add = x + 3

# Scalar subtraction
x_sub = x - 2

print("Original :", x)
print("After +3 :", x_add)
print("After -2 :", x_sub)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))

plt.plot(x, marker='o', linewidth=2, label='Original')
plt.plot(x_add, marker='s', linewidth=2, label='After +3')
plt.plot(x_sub, marker='^', linewidth=2, label='After -2')

plt.title("Effect of Scalar Addition and Subtraction")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.legend()
plt.show()
```
